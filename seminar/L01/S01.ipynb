## 3. Compute statistics
Compute the sample mean and the sample covariance of the generated data.

```python
def sample_mean(dat):
    # Your code below
    s_mean = 1. / dat.shape[0] * np.sum(dat)
    return s_mean

def smean_iscorrect(dat):
    return np.isclose(sample_mean(dat), dat.mean(), rtol=1e-14)
```

Check correctness of your `sample_mean(...)`

```python
sample_mean(r_set)
```

```python
smean_iscorrect(r_set)
```

```python
def sample_covariance(dat):
    # Your code below
    s_cov = 1. / (dat.shape[0] - 1) * np.sum((dat - dat.mean()) * (dat - dat.mean()).T)
    return s_cov

def scov_iscorrect(dat):
    print(np.cov(dat, bias=False))
    return np.isclose(sample_covariance(dat), np.cov(dat, bias=False), rtol=1e-14)
```

```python
sample_covariance(r_set)
```

```python
scov_iscorrect(r_set)
```

If we change the bias term (1/(N-1)) in covariance to 1/N then for the smaller amount of samples the change is dramatical.
